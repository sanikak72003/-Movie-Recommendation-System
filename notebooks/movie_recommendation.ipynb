## extracting specific words from genre,crew,cast by ignoring id

```python
# our genre,cast,crew cols are still a list of dictionaries where each dictionary has id and name as keys and id val is an int which 
# we dont need but name is a string that we do need
```

```python
# function to extract just the name part from the dictionary from inisde the list of dictionaries

def extract_name(lst):

    L=[]

    for dict in lst:

        L.append(dict['name'])

    return L
```

```python
movies['genres']=movies['genres'].apply(extract_name)

# type(movies['genres'].iloc[0])

# when you apply extract_name function on col genre it iterates through every cell (row) in that col
# every record contains a list of dictionaries where each dictionary has id and name of the genre
# so the lst is passed to the function
# create an empty list called L that will store the extracted names
# now dict obj will traverse every dictionary in lst and extract the name of genre using the key name and append the value in L 
# at last return L

# for each record which is a list of dictionaries the method/function processes it and retrns a list of names of the genres that were part of diff 
# dictionaries
```

```python
movies['genres'].iloc[0]
```

```python
# do the same for keywords 

movies['keywords']=movies['keywords'].apply(extract_name)

movies['keywords'].iloc[0]
```

```python
movies['cast'].iloc[0][0]
# cast col each record contains the cast present in that movie
# it is a list of dictiionaries
# each dictionary has cast id,name of character,name of person playing that character etc
# here we extract 0th row first dictionary from the cast col
```

```python
# now for the cast extract only the top 3 names (name of person plating role not character name)


def extract_top3_names(lst):

    L=[]

    for dict in lst[0:3]:
        L.append(dict['name'])

    return L

    
```

```python
# now perform this stemming on each word of each str present in the tag col

def stem(str):

    y=[] # empty list will store the stemmed words later this list will be converted to string using join()

    for word in str.split():  # str.split() splits the string into a list and word visits every element of that list which are words
        y.append(ps.stem(word))  # stem every word and store in list y

    return ' '.join(y) # join joins all the elements in list y and seperate them by space leads to formation of string which is returned back to tags col
    # now tags col will contain a str with every word stemmed
        
```

```python
movies['tags']=movies['tags'].apply(stem) # tags col each record/row contains a string

# stem function takes that string
# assign empty list to store stemmed words
# split the string into a list and visit every word in that list
# stem that word and store in list y
# convert list y to string and return back to particulat record/row of tags col
```

```python
movies['tags'].iloc[0]
```

```python
from sklearn.feature_extraction.text import CountVectorizer

cv=CountVectorizer(max_features=5000,stop_words='english')
# max_features=5000 means select atmost 5000 vocabulary words based on how often they occur by checking the tags of all the movied
# stop_words means ignore common english words like a an the etc

vectors=cv.fit_transform(movies['tags']).toarray()
# cv.fit reads the every record/row of tags col which contains a string containing all words related to the movie
# so it reads the tags associated with all movies and selects the best 5000 words for further processing
# transform creates features for those vocabulary words and then count the freq of them in each movie and assign col value accordingly

# cv.fit_transform() by default creates a sparse matrix that is such that the unnecessary space occupied by zeros is reduced but convert to normL NUMPY
# array using toarray() and store in vectors
```

```python
vectors[0]  # stores the array containing the count of words selected as features if present in the movie description
# eg let say movie description is hina is a good girl and she likes to go to the gym and she likes is a coffee girl as she loves coffee
# let say the feature selected are good girl gym coffee
# for the above description the vector will be like [1,2,1,2]  # good appears once and girl and coffee appear twice
```

```python
# now check similarity between the movies
# now every movie is a vector inside the vectors array

# suppose vector A is [1,1,0,0] and vector B is [0,1,1,0] then similarity(A,B)=A.B/||A|| * ||B|| (dot product of two vectors/ product of vector magnitudes)
# why we use cosine similarity and not just count similar words ? --> let say movie A description has 100 words and B description has 150 words
# both have a lot of similar words but both description have diff length so instead we consider direction

# SEE REASON PROPERLY ****
```

```python
from sklearn.metrics.pairwise import cosine_similarity

sim=cosine_similarity(vectors)
```

## final combination

```python
# now we have our movies dataset
# vector represemnting the movies and the features(vocabulary) associated with them
# similarity score of each movie with each other movie
# combine them all

# if user enters the name of a movie retrieve its row index from the movies dataset by comparing the movie name with records/rows of original_title col
# once the row index is retrieved go to that index in the sim vector that has similarity of every movie with every other movie
# inside that sim vector at the given row index extract the entire row
# sort the values in desceding order so that most similar movies appear first
# select top 5 similar movies for recommendation
```

```python
def recommend(movie_name):  # pass name of movie to this function

    # retrieve row index of that movie
    row_idx=movies[movies['original_title']==movie_name].index[0]

    # now select the row from sim that contains similarity score of given movie with other movies

    sim_scores=sim[row_idx]  # get the entire row of similarity score where at idx i there is similarity of curr movie with movie at row i in 
    # original movies dataset
    # so maintain the indices of the similarity scores
    # if you simply sort them you will loose which movie at which idx

    sim_scores_lst=list(enumerate(sim_scores)) # list of tuples whwrre each tuple has idx and sim score at that idx
    # the idx is same as idx of the movie in original movies dataset and its sim score with current movie

    # sort on the basis of similarity score keeping the idx associated with it constant

    sorted_sim_scores=sorted(
        sim_scores_lst,  # pass the list of tuples eg [(0,1),(1,0.7),(2,0.54),(3.0.9)] # 0 , 1 , 2 , 3 are indices of movies and the
        # second number are their similarity scores with curr movie
        key=lambda x:x[1],  # x traverses every element in sim_scores_lst i.e x is a tuple and select the second element of that tuple i.e similarity score
        # sort on basis of key i.e similarity score
        reverse=True  # sort in descending order so set reverse as true
    )
    # result is stored in sorted sim scores
    # for above eg it would be [(0,1),(3,0.9),(1,0.7),(2,0.54)]

    # extract top 5

    top_5=[]
    

    for i in range(1,6): #start from idx1 as at idx0 there will be the same movie with idx0 and as its sim score with iself is max i.e 1 so it will appear
        # at idx0 in list

        idx=sorted_sim_scores[i][0] # one by one visit the first 5 elements of sorted_sim_score list and at each element which is atuple extract its first elem
        # which is idx of movie

        # print the name of movie at that idx from movies dataset


        top_5.append(movies['original_title'].iloc[idx])

    return top_5

    
```
